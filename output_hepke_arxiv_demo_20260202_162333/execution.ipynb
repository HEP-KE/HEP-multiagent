```
import json

# Load the paper metadata
file_path = "" + REPLAY_DIR + "/dark_matter_papers_metadata.json"

with open(file_path, 'r') as f:
    papers_data = json.load(f)

# Extract papers list
papers = papers_data['papers']

print("All papers with years:")
for i, paper in enumerate(papers, 1):
    print(f"{i}. {paper['title']} ({paper['year']})")

print("\n" + "="*50)

# Create year count dictionary
year_counts = {}
papers_by_year = {}

for paper in papers:
    year = paper['year']
    title = paper['title']
    
    # Count papers by year
    if year in year_counts:
        year_counts[year] += 1
    else:
        year_counts[year] = 1
    
    # Store papers by year for reference
    if year not in papers_by_year:
        papers_by_year[year] = []
    papers_by_year[year].append({
        'title': title,
        'arxiv_id': paper['arxiv_id'],
        'authors': paper['authors']
    })

print("Year counts:")
for year in sorted(year_counts.keys()):
    print(f"{year}: {year_counts[year]} paper(s)")

print("\nPapers by year:")
for year in sorted(papers_by_year.keys()):
    print(f"\n{year}:")
    for paper in papers_by_year[year]:
        print(f"  - {paper['title']}")
        print(f"    ID: {paper['arxiv_id']}")
        print(f"    Authors: {paper['authors']}")

# Create the final data structure for JSON export
publication_year_data = {
    'year_counts': year_counts,
    'papers_by_year': papers_by_year,
    'total_papers': len(papers),
    'year_range': f"{min(year_counts.keys())}-{max(year_counts.keys())}",
    'analysis_summary': {
        'unique_years': len(year_counts),
        'most_productive_year': max(year_counts.keys(), key=year_counts.get),
        'papers_in_most_productive_year': max(year_counts.values())
    }
}

print(f"\n" + "="*50)
print("ANALYSIS SUMMARY:")
print(f"Total papers: {publication_year_data['total_papers']}")
print(f"Year range: {publication_year_data['year_range']}")
print(f"Unique years: {publication_year_data['analysis_summary']['unique_years']}")
print(f"Most productive year: {publication_year_data['analysis_summary']['most_productive_year']}")
print(f"Papers in most productive year: {publication_year_data['analysis_summary']['papers_in_most_productive_year']}")
```
