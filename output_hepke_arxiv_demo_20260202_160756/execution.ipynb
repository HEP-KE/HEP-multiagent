```
import json

# Load the JSON data
file_path = "" + REPLAY_DIR + "/dark_matter_papers.json"

with open(file_path, 'r') as f:
    papers_data = json.load(f)

papers = papers_data['papers']

# Extract publication years and create year distribution
year_distribution = {}

print("Extracting publication years...")
for paper in papers:
    year = paper.get('year')
    if year is not None:
        if year in year_distribution:
            year_distribution[year] += 1
        else:
            year_distribution[year] = 1
        print(f"  {paper['title'][:50]}... -> {year}")

print(f"\nYear distribution:")
for year in sorted(year_distribution.keys()):
    print(f"  {year}: {year_distribution[year]} paper(s)")

print(f"\nSummary:")
print(f"Total papers processed: {sum(year_distribution.values())}")
print(f"Year range: {min(year_distribution.keys())} - {max(year_distribution.keys())}")
print(f"Span: {max(year_distribution.keys()) - min(year_distribution.keys())} years")

# Create the final data structure for saving
year_data = {
    "year_distribution": year_distribution,
    "metadata": {
        "total_papers": sum(year_distribution.values()),
        "year_range": {
            "earliest": min(year_distribution.keys()),
            "latest": max(year_distribution.keys())
        },
        "span_years": max(year_distribution.keys()) - min(year_distribution.keys()),
        "source_file": "dark_matter_papers.json",
        "analysis_type": "publication_year_distribution"
    }
}

print(f"\nFinal data structure:")
print(f"Year distribution: {year_data['year_distribution']}")
print(f"Metadata: {year_data['metadata']}")
```
